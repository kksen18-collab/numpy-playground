```python
# numpy is conventionally imported as np
import numpy as np
```

```python
# default_rng is the modern generator API - unseeded, so results vary per run
rng = np.random.default_rng()
```

```python
# random integers as a 3x2 array; low is included, high is excluded
print(rng.integers(low=10, high=69, size=(3,2)))
```

```python
# passing a seed makes the sequence reproducible - same numbers every run
rng_seed = np.random.default_rng(42)

print(rng_seed.integers(low=10, high=69, size=(3,2)))
```

```python
# legacy np.random API: one random float in [0.0, 1.0)
print(np.random.uniform())
```

```python
# legacy global seed, then random floats in [10.0, 69.0) as a 3x2 array
np.random.seed(19)
print(np.random.uniform(low=10, high=69, size=(3,2)))   
```

```python
# shuffle reorders the array in place and returns None
rng = np.random.default_rng()

array = np.array([1, 2, 3, 4, 5])
rng.shuffle(array)
print(array)
```

```python
# pick a single random element out of the array
fruits = np.array(['apple', 'banana', 'cherry', 'date', 'elderberry'])
fruit = rng.choice(fruits)
print(fruit)
```

```python
# pick 3 elements - with replacement by default, so repeats are possible
fruit = rng.choice(fruits,size=3)
print(fruit)
```

```python
# size can be a shape, giving a (1, 4) array rather than a flat one
fruit = rng.choice(fruits,size=(1,4))
print(fruit)
```
